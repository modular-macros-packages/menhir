```python
global directory
directory = "../_build/default/data/benchmark-GLR"
def slope(df, x, y):
  model = LinearRegression().fit(df[x].to_frame(), df[y].to_frame())
  slope = model.coef_[0][0]
  return("{:s} = O({:s}^{:.2f})".format(y, x, slope))
  
def grammar(name):
  # TODO increase dpi
  # Read the data.
  df = pd.read_csv("{:s}/{:s}.csv".format(directory, name))
  # Move to log-log scale.
  logdf=np.log10(df)
  # Draw nodes and time as functions of input size.
  g = sns.PairGrid(logdf, y_vars=["nodes", "time", "rejects"], x_vars=["n"], height=3)
  # In every box, draw a linear regression plot.
  g.map(sns.regplot)
  # Show the slope of nodes versus n.
  print(slope(logdf, "n", "nodes"))
  # Show the slope of time versus n.
  print(slope(logdf, "n", "time"))
  # Show the slope of rejects versus n.
  print(slope(logdf, "n", "rejects"))
```
